```python
import mysql.connector as connector
connection = connector.connect(user = "root", password = "rootpass", db = "littlelemondb")
cursor = connection.cursor()
```

# Queries

```python
show_tables_query = "SHOW tables" 
cursor.execute(show_tables_query)
results = cursor.fetchall()
print(results)
```

```python
join_query = """SELECT Customers.CustomerName, Customers.Email, Customers.Phone, Orders.TotalCost
FROM Customers INNER JOIN Orders ON Customers.OrderID = Orders.OrderID
WHERE Orders.TotalCost > 60;"""
cursor.execute(join_query)
results = cursor.fetchall()
print(results)
```
